# 02 - Split en Train y en Test

## Objetivo

Una vez que tenemos filtrados y balanceados las imagenes a utilizar para el desarrollo del modelo, deberemos dividir estas imagenes en **train** (las imagenes que van a ser utilizadas para el entrenamiento del modelo) y **test** (Las imagenes utilizadas para comprobar la calidad del modelo). 

## 1. Importacion de librerias y carga de la muestra

Importo las librerias necesarias, y realizo la carga del data set ya filtrado y balanceado, como se realizo en el paso anterior.

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split

PROCESSED = Path("../data/processed")

df = pd.read_csv(PROCESSED / "metadata.csv")
print(len(df))
print(df["dx"].value_counts())
```

## 2. Cantidad de imagen para entrenar

Se define que se va a utilizar un 80% de imagenes para el entrenamiento del modelo y un 20% para medir el recall del modelo.

```python

train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    stratify=df["etiqueta"],
    random_state=42,
)

train_df.to_csv(PROCESSED / "train.csv", index=False)
test_df.to_csv(PROCESSED / "test.csv", index=False)
```
